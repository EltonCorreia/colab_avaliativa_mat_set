Parte D – Resto da divisão e divisão inteira

19. Uma turma possui 47 alunos e será dividida em grupos de 6. Quantos grupos completos podem
ser formados e quantos alunos sobram? Utilize // e %.
20. Uma empresa possui 1.253 unidades de determinado produto e deseja armazená-las em caixas
com capacidade para 24 unidades. Determine o número de caixas completamente preenchidas e a
quantidade que ficará fora delas.

```python
alunos = 47
gp_alunos = 6

grupos = alunos//gp_alunos
sobra = alunos%gp_alunos

print(f"Haverá {grupos} grupos exatos e restarão {sobra} alunos.")
```

```python
unidades = 1253
unidades_caixas = 24

caixas = unidades//unidades_caixas
sobra = unidades%unidades_caixas

print(f"Haverá {caixas} caixas exatas preenchidas e sobrará {sobra} unidades fora.")
```

24. Velocidade média: uma motocicleta percorre 286 km em 3,5 horas. Calcule a velocidade média

Parte F – Utilizando for e range()

27. Utilize for para imprimir os números inteiros de 1 até 20, um por linha.
28. Utilize for para imprimir os quadrados dos números de 1 até 10. Exemplo: 1² = 1, 2² = 4, etc.
29. Utilize for para imprimir as potências de 2 de 0 até 10. O resultado deve aparecer organizado,
indicando o expoente e o valor
30. Utilize for para imprimir os múltiplos de 7 entre 7 e 70.
31. Utilize for para imprimir os números de 1 a 30 e informar, para cada número, se ele é par ou
ímpar. Utilize % para fazer a verificação.

```python
for i in range(1,21):
  print(i)
```

```python
for i in range(1,11):
  print(f"{i}² = {i**2}")
```

```python
for i in range(11):
  q = 2**i
  print(f"{i}² = {q}")
```

```python
for i in range(7,71,7):
  print(i)
```

```python
for i in range (1,31):
  if i%2==0:
    print(f"{i} é par.")
  else:
    print(f"{i} é ímpar.")
```
